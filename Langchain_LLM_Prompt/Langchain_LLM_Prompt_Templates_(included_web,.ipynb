how to create multiple AI tools using LangChain and Groq, where each tool performs a specific task by using a different prompt and the same LLM.

Langchain_LLM_Prompt_Template_MutipleTool

```python
from google.colab import userdata

from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.tools import tool

# ============================================================
# Load Groq API Key from Google Colab Secrets
# Secret Name: GROQ_API_KEY
# ============================================================

GROQ_API_KEY = userdata.get("GROQ_API_KEY")

# ============================================================
# Initialize Groq LLM
# ============================================================

llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    groq_api_key=GROQ_API_KEY,
    temperature=0
)

# ============================================================
# Simple QA Prompt
# ============================================================

qa_prompt = ChatPromptTemplate.from_template(
    "Answer clearly:\n\nQuestion: {question}"
)

qa_chain = qa_prompt | llm

# ============================================================
# Summarizer Prompt
# ============================================================

summary_prompt = ChatPromptTemplate.from_template(
    "Summarize the following text:\n\n{text}"
)

summary_chain = summary_prompt | llm

# ============================================================
# Tool 1 : Simple QA
# ============================================================

@tool
def simple_qa(question: str) -> str:
    """Answer factual questions clearly."""
    response = qa_chain.invoke({"question": question})
    return response.content

# ============================================================
# Tool 2 : Summarizer
# ============================================================

@tool
def summarizer(text: str) -> str:
    """Summarize the given text."""
    response = summary_chain.invoke({"text": text})
    return response.content

# ============================================================
# Example Inputs
# ============================================================

qa_query = "What is LangGraph in LangChain?"

summary_text = """
LangGraph is a library built on top of LangChain that helps developers create
stateful, multi-step AI applications as graphs.
Each node represents a task such as calling an LLM,
executing a tool, or making a routing decision.
It is ideal for complex AI workflows, multi-agent systems,
and applications that require memory and branching logic.
"""

# ============================================================
# Run Tools
# ============================================================

qa_answer = simple_qa.invoke(qa_query)
summary_answer = summarizer.invoke(summary_text)

# ============================================================
# Display Output
# ============================================================

print("=" * 80)
print("🧠 SIMPLE QA TOOL")
print("=" * 80)
print(qa_answer)

print("\n" + "=" * 80)
print("📝 SUMMARIZER TOOL")
print("=" * 80)
print(summary_answer)
```

```python
from google.colab import userdata
import os

from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.tools import tool
from langchain_community.tools.tavily_search import TavilySearchResults

# ============================================================
# Load API Keys
# ============================================================

GROQ_API_KEY = userdata.get("GROQ_API_KEY")
TAVILY_API_KEY = userdata.get("TAVILY_API_KEY")

os.environ["TAVILY_API_KEY"] = TAVILY_API_KEY

# ============================================================
# Initialize Groq LLM
# ============================================================

llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    groq_api_key=GROQ_API_KEY,
    temperature=0
)

# ============================================================
# QA Prompt
# ============================================================

qa_prompt = ChatPromptTemplate.from_template("""
Answer clearly and accurately.

Question:
{question}
""")

qa_chain = qa_prompt | llm

# ============================================================
# Summarizer Prompt
# ============================================================

summary_prompt = ChatPromptTemplate.from_template("""
Summarize the following text:

{text}
""")

summary_chain = summary_prompt | llm

# ============================================================
# Tool 1 : Simple QA
# ============================================================

@tool
def simple_qa(question: str) -> str:
    """Answer factual questions clearly."""
    response = qa_chain.invoke({"question": question})
    return response.content

# ============================================================
# Tool 2 : Summarizer
# ============================================================

@tool
def summarizer(text: str) -> str:
    """Summarize the given text."""
    response = summary_chain.invoke({"text": text})
    return response.content

# ============================================================
# Tool 3 : Tavily Web Search
# ============================================================

search_tool = TavilySearchResults(max_results=3)

# ============================================================
# Example Inputs
# ============================================================

qa_query = "What is LangGraph in LangChain?"

summary_text = """
LangGraph is a framework built on top of LangChain for creating
stateful AI workflows. It represents workflows as graphs where
nodes perform tasks, edges define execution flow, and state
stores shared information between steps.
"""

search_query = "Latest updates on GPT-4o by OpenAI"

# ============================================================
# Run Tools
# ============================================================

qa_answer = simple_qa.invoke(qa_query)

summary_answer = summarizer.invoke(summary_text)

search_answer = search_tool.invoke(search_query)

# ============================================================
# Display Results
# ============================================================

print("=" * 80)
print("🧠 SIMPLE QA TOOL")
print("=" * 80)
print(qa_answer)

print("\n" + "=" * 80)
print("📝 SUMMARIZER TOOL")
print("=" * 80)
print(summary_answer)

print("\n" + "=" * 80)
print("🌐 TAVILY WEB SEARCH TOOL")
print("=" * 80)
print(search_answer)
```

```python
from google.colab import userdata
import os

from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langchain_tavily import TavilySearch

# ============================================================
# API Keys
# ============================================================

GROQ_API_KEY = userdata.get("GROQ_API_KEY")
TAVILY_API_KEY = userdata.get("TAVILY_API_KEY")

os.environ["TAVILY_API_KEY"] = TAVILY_API_KEY

# ============================================================
# Groq Model
# ============================================================

llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    groq_api_key=GROQ_API_KEY,
    temperature=0
)

# ============================================================
# Web Search Tool
# ============================================================

search_tool = TavilySearch(max_results=3)

# ============================================================
# Create Agent
# ============================================================

agent = create_agent(
    model=llm,
    tools=[search_tool],
    system_prompt="""
You are a helpful AI assistant.

Rules:
- Answer directly if you already know the answer.
- Use the Tavily search tool ONLY when the user asks for:
    * latest news
    * current events
    * today's information
    * live information
    * recent updates
- Be concise and accurate.
"""
)

# ============================================================
# Queries
# ============================================================

queries = [
    "What is LangGraph in LangChain?",
    "Latest updates on GPT-4o by OpenAI",
]

# ============================================================
# Run
# ============================================================

for query in queries:

    print("=" * 80)
    print("USER:")
    print(query)

    result = agent.invoke(
        {
            "messages": [
                {
                    "role": "user",
                    "content": query
                }
            ]
        }
    )

    print("\nAGENT RESPONSE:\n")
    print(result["messages"][-1].content)
```
